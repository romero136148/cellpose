# Batch segmentation with image restoration

This notebook shows how to run Cellpose with an image restoration filter on a directory of images. The code mirrors the command line usage from the documentation:
```bash
python -m cellpose --dir /path/to/images --model_type cyto3 --restore_type denoise_cyto3 --diameter 25 --chan2_restore --chan 2 --chan2 1
```

Activate your `cellpose` environment (e.g. `source cellpose/bin/activate`) before running so the package is available.


````
from cellpose import io, denoise
import glob
from pathlib import Path

# directory containing PNG files
img_dir = Path('/path/to/images')
img_paths = sorted(glob.glob(str(img_dir / '*.png')))

# load images as a list of numpy arrays
imgs = [io.imread(p) for p in img_paths]
print(f'Loaded {len(imgs)} images')
````

````
# initialize combined restoration + segmentation model
model = denoise.CellposeDenoiseModel(
    gpu=True,
    model_type='cyto3',
    restore_type='denoise_cyto3',
    chan2_restore=True
)
````
